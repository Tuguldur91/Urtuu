## Mini-course: computing fundamentals for a team quiz

**What the guideline confirms:** Finale is a question-and-answer competition worth 25 points. It does not specify subjects. These are useful foundations for a broad computing quiz.

### Lesson 1 — Programming and data structures
A variable names a value; a condition selects a branch; a loop repeats steps. A Boolean is true or false. `and` requires both conditions, `or` requires at least one, and `not` reverses truth. A list stores a sequence. A stack removes the newest item first (LIFO); a queue removes the oldest first (FIFO). A dictionary maps keys to values.

### Lesson 2 — Algorithms and growth
An algorithm is a sequence of steps solving a problem. Big-O describes how work grows with input size. O(1) is constant work, O(n) grows linearly, and O(n²) grows quadratically. Binary search repeatedly halves a sorted search range and takes O(log n) steps. Sorting is a prerequisite; applying it to arbitrary unsorted data can give a wrong answer.

### Lesson 3 — Networks and operating systems
A local network connects nearby machines; the internet connects networks globally. Local services can operate without internet access. IP addresses identify network interfaces; DNS looks up records for names. An operating system manages resources and processes. A process is a running program. Concurrency allows tasks to make progress over overlapping periods; parallelism means simultaneous execution.

### Lesson 4 — Databases and representation
A database table stores rows and columns. A primary key uniquely identifies a row. SQL `SELECT` retrieves data, `WHERE` filters rows, and `COUNT(*)` counts rows. Encoding changes representation; encryption uses a key to protect content; a cryptographic hash produces a fixed-size digest used to compare data. A hash cannot simply be decoded back into a file.

## Quiz answers

| # | Answer |
|---|---|
| 1 | 45 = 32 + 8 + 4 + 1 |
| 2 | The requested resource was not found |
| 3 | The server refuses access to the resource |
| 4 | Queue |

| 6 | O(n) |
| 7 | To uniquely identify rows; primary keys must be unique and non-null |

**Lightning answers:** (1) No; concurrency can use interleaving on one core. (2) Yes, if local addressing, connectivity, and permissions allow it. (3) The number of matching rows, not their score sum. (4) Accounts, devices, and IPs may be shared or compromised; correlate session and audit evidence.

```python
import base64, hashlib
parts = sorted(chunks, key=lambda item:item["part"])
assert [p["part"] for p in parts] == list(range(1,7)), "Missing or duplicate parts"
restored = b"".join(base64.b64decode(p["data"],validate=True) for p in parts)
actual = hashlib.sha256(restored).hexdigest()
print(restored.decode("utf-8"))
print("SHA-256:", actual)
assert actual == expected_sha256
print("Restoration verified.")
```
