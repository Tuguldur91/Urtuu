## JavaScript mini-course for Glitch

JavaScript is a plausible implementation language for a web-interface station, but the guideline does not confirm the language or source-code access. These examples teach both what to test on screen and what to inspect if code is available.

### 1. Values and conversion

`const` declares a binding that cannot be reassigned; it does not freeze an object or array. `let` allows reassignment. Strings use quotes, numbers do not, and booleans are `true` or `false`. Prefer `===` when you want equality without type conversion. An input's `.value` is a string: `"4" + 1` gives `"41"`, while `Number("4") + 1` gives `5`. Conversion is not validation: `Number("")` is 0, so check empty text separately. `Number.isFinite(value)` rejects `NaN` and infinity; `Number.isInteger(value)` checks for a whole number.

### 2. Conditions and UI state

`if (value)` checks truthiness, not whether text spells “true.” Both `"false"` and `"0"` are truthy because they are nonempty strings. `false`, `0`, `""`, `null`, `undefined`, and `NaN` are falsy. A checkbox's `.checked` is a boolean; its `.value` is the text to send when selected. A checkbox without an explicit value normally has `.value === "on"` even when unchecked.

### 3. Arrays and callbacks

`filter` keeps elements for which its callback returns a truthy value. `map` builds an array from callback results. `x => x > 0` returns its expression, but `x => { x > 0; }` returns `undefined`: braces need an explicit `return`. `sort()` mutates the array and compares strings by default; numeric sorting needs a comparator such as `(a, b) => a - b`.

### 4. DOM and events

The DOM is the browser's object representation of the page. `document.querySelector(...)` finds the first matching element or returns `null`. `.textContent` changes displayed text. `addEventListener("click", handler)` registers a function to run later; `handler()` calls it immediately. Forms also have default browser behavior: submitting usually navigates or reloads. Validate on the form's `submit` event so clicking and pressing Enter follow the same path. When JavaScript handles submission, `event.preventDefault()` stops the browser's default submission; it does not itself save the data.

### 5. Scope and delayed work

A closure lets a function refer to variables in the surrounding scope. `var` in a loop shares one binding across callbacks; `let` creates a binding for each loop iteration. A callback that runs later can therefore see a different value than the one you expected when registering it.

### 6. Promises, fetch, and request order

A Promise represents a result that may arrive later. `await` waits for that result inside an `async` function. `fetch` normally resolves even when the HTTP response is 404 or 500; check `response.ok` before displaying success. A rejected request needs error handling too. Two requests may finish in the opposite order from which they started: track which result is still relevant before rendering it.

**Debugging routine:** reproduce the visible failure → inspect values/types → trace the event or response order → make one focused fix → repeat the original case and a nearby boundary case. If no code is exposed, the same reasoning still helps you choose inputs and write useful bug reports.

### JS7 — Check HTTP outcome and catch rejection

The broken code reports `Saved`. Awaiting a response is not evidence of HTTP success. Check `ok` and provide a failure path. This minimal example assumes a successful HTTP response is sufficient; an application may also require checking its response body.

```javascript
{
  async function save(fetcher) {
    try {
      const response = await fetcher();
      if (!response.ok) throw new Error(`HTTP ${response.status}`);
      return "Saved";
    } catch (error) {
      return `Save failed: ${error.message}`;
    }
  }
  save(async () => ({ ok: false, status: 404 })).then(console.log);
  // Save failed: HTTP 404
  save(async () => ({ ok: true, status: 200 })).then(console.log);
  // Saved
  save(async () => { throw new Error("Network unavailable"); }).then(console.log);
  // Save failed: Network unavailable
}
```

### JS8 — Ignore stale responses

The broken code ends with `Results for ca`. The older request finishes last and overwrites the current result. Assign each request a sequence number and render only if it still matches the latest request. The guard also belongs on loading/error UI updates in a complete application.

```javascript
{
  const pending = new Map();
  const fakeSearch = query => new Promise(resolve => pending.set(query, resolve));
  let displayed = "";
  let latestRequest = 0;
  async function search(query) {
    const request = ++latestRequest;
    const result = await fakeSearch(query);
    if (request === latestRequest) displayed = result;
  }
  const older = search("ca");
  const newer = search("cat");
  pending.get("cat")("Results for cat");
  pending.get("ca")("Results for ca");
  Promise.all([older, newer]).then(() => console.log(displayed));
  // Results for cat
}
```

**Turn a fix into a report:** “With query `cat` visible in the input, the results switch back to `ca` when its older response arrives. Expected: results remain associated with `cat`. Reproduce by resolving the newer request first. The fix ignores responses whose request number is no longer current.”
