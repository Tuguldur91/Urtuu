## Mini-course: finding and explaining interface bugs

**What the guideline confirms:** Find user-interface errors. Source-code access, developer tools, and the exact bug types are unspecified.

### Lesson 1 — Expected versus observed behavior
A bug is a mismatch between the required behavior and what happens. Read requirements first. “Exactly four members” gives useful tests: 3, 4, and 5. This is boundary testing: test around the point where behavior should change. Also try an empty field, spaces only, and repeated values.

### Lesson 2 — What a webpage consists of
HTML describes elements such as inputs and buttons. CSS controls appearance. JavaScript can react to clicks and update data. A button can look correct but perform the wrong action. A form can show an error correctly but still submit invalid data. Check the complete interaction.

### Lesson 3 — Common interface checks
Test navigation, submission, reset, validation messages, scrolling, and keyboard focus. Check whether a label identifies its input. Look for unreadable contrast, overlapping text, hidden controls, and inconsistent counts. Distinguish serious failures from cosmetic issues.

### Lesson 4 — Report something another person can reproduce
Record the starting state, exact input, steps, expected result, actual result, and impact. Example: “On page 1 with ten teams, advance through all pages. Teams 4 and 8 never appear. Expected: all ten teams appear once.” Python slices exclude the ending index: `rows[0:4]` returns indices 0, 1, 2, 3.

## Worked example: reproducible bug reports
**Requirement:** Four members from at least two programs.
**Steps:** Enter four members, all from one program; submit.
**Expected:** Reject registration and explain the program requirement.
**Observed:** Registration accepted.
**Impact:** An ineligible team can register.

Use this format instead of “the form is broken.” Prioritize task-blocking errors over cosmetic problems. Only inspect the provided practice environment.

## Your exercises

```python
def valid_registration(name, members, programs):
    return bool(name) and members >= 4 and len(programs) >= 2

teams = [f"Team {i}" for i in range(1,11)]
def page_rows(rows, page, size=4):  # page is zero-based
    start = page*size
    return rows[start:start+size-1]

html = """<form>
<label for="email">Email</label><input id="mail" type="text">
<label for="count">Members (must be four)</label><input id="count" type="number" min="1" max="8">
<button type="reset">Submit registration</button>
<img src="status.png">
<p style="color:#eeeeee;background:white">Registration closes November 10.</p>
</form>"""
print(html)
# Your reports and fixes here
```

```python
def corrected_registration(name, members, programs):
    return bool(name.strip()) and members == 4 and len(set(programs)) >= 2

cases = [("   ",4,["CS","IT"]), ("A",5,["CS","IT"]), ("A",4,["CS","CS"])]
for c in cases:
    print(c, "old:", valid_registration(*c), "fixed:", corrected_registration(*c))

def corrected_page_rows(rows, page, size=4):
    return rows[page*size:(page+1)*size]
print("Buggy pages:", [page_rows(teams,p) for p in range(3)])
print("Fixed pages:", [corrected_page_rows(teams,p) for p in range(3)])
assert sum((corrected_page_rows(teams,p) for p in range(3)), []) == teams
```

**Exercise 1:** Whitespace passes `bool(name)`; five members pass `>= 4`; duplicate program names pass a raw list-length check. The corrected function assumes program labels are already normalized. Production validation should also check data types and the number of member records.

**Exercise 2:** Teams 4 and 8 disappear because the slice endpoint excludes one row per page.

**Exercise 3:** The email label points to a missing ID; email uses a text field instead of email validation; the member limits allow ineligible counts; the “Submit” button resets the form; the image lacks alternative text (use descriptive alt text if meaningful, empty alt if decorative); the deadline text has very poor contrast. Frontend restrictions also need server-side validation.
