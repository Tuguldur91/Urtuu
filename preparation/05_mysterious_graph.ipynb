## Mini-course: graph puzzles from the beginning

**What the guideline confirms:** Solve an “interesting graph” problem. No particular algorithm is named.

### Lesson 1 — What a graph means
A graph contains vertices (nodes) and edges (connections). Rooms can be nodes and corridors edges. A directed edge has an arrow and permits only the indicated direction. An undirected edge works both ways. A weight is a cost such as distance or time. A path follows connected vertices. A cycle returns to its start. A node’s degree in a simple undirected graph is the number of incident edges.

### Lesson 2 — Match the question to the algorithm
BFS (breadth-first search) explores one distance layer at a time using a queue: first in, first out. It finds the fewest edges in an unweighted graph. DFS (depth-first search) follows one branch before backtracking; it is useful for exploring connectivity. Keep a visited set so cycles do not make you loop forever.

Dijkstra finds the lowest total weight when weights are nonnegative. Track the best known distance to each node. Repeatedly take the node with the smallest tentative distance and try improving its neighbors. A priority queue efficiently selects that smallest distance. Fewest edges and lowest cost can give different answers.

### Lesson 3 — Visiting edges versus vertices
An Euler trail uses every edge exactly once. For an undirected graph whose non-isolated vertices are connected, it exists when there are zero or two odd-degree vertices. With two, start at one odd vertex and finish at the other. A Hamiltonian path instead visits every vertex exactly once; the Euler degree rule does not solve that problem.

### Lesson 4 — Coloring
A proper coloring gives adjacent vertices different colors. A triangle needs three colors. A graph is bipartite if two colors suffice; an odd cycle prevents this. Try assigning alternating colors while traversing the graph and check every edge for conflicts.

```python
from collections import deque
def bfs(start, target):
    q=deque([(start,[start])]); seen={start}
    while q:
        node,path=q.popleft()
        if node == target: return path
        for nxt,_ in graph[node]:
            if nxt not in seen:
                seen.add(nxt); q.append((nxt,path+[nxt]))
print("BFS:", bfs("A","F"))

import heapq
def dijkstra(start,target):
    q=[(0,start,[start])]; best={start:0}
    while q:
        distance,node,path=heapq.heappop(q)
        if distance != best[node]: continue
        if node == target: return distance,path
        for nxt,weight in graph[node]:
            candidate=distance+weight
            if candidate < best.get(nxt,float("inf")):
                best[nxt]=candidate
                heapq.heappush(q,(candidate,nxt,path+[nxt]))
print("Dijkstra:", dijkstra("A","F"))
print("Degrees:", {n:len(adj) for n,adj in graph.items()})
colors={"A":0,"B":1,"C":2,"D":0,"E":1,"F":2}
assert all(colors[a] != colors[b] for a,b,_ in edges)
print("Valid coloring:",colors)
```

**Answers:** A–B–D–F is a 3-hop route; A–C–E–F also has 3 hops. No 2-hop route exists because A’s neighbors B and C are not adjacent to F. The cheapest cost is 8: A–B–C–E–F. A–B–D–E–F costs 9, so it is not another optimum. The minimum-cost route is unique in this graph.

Degrees are A=2, B=3, C=3, D=3, E=3, F=2. The graph is connected but has four odd-degree vertices, so no Euler trail exists. A connected undirected graph permits an Euler trail only with zero or two odd-degree vertices. Triangle A–B–C prevents two-coloring.
