## Mini-course: from unfamiliar symbols to a readable clue

**What the guideline confirms:** Decode a message and find something hidden in the room. It does not name the cipher.

### Lesson 1 — Characters, bytes, and number bases
A character is a symbol such as `A`. Computers store data as numbers in bytes; one byte contains eight bits. A bit is 0 or 1. Binary uses place values 128, 64, 32, 16, 8, 4, 2, 1. Thus `01000001` is 64 + 1 = 65, which represents `A` in ASCII. Hexadecimal is a shorter way to write the same bytes: `41` means 4 × 16 + 1 = 65. Hex digits A–F mean 10–15.

**Try by hand:** What character is binary `01000010`, or hex `42`? Both are decimal 66: `B`. ASCII covers basic English characters; Mongolian text normally needs a broader encoding such as UTF-8, where a character can occupy several bytes.

### Lesson 2 — Encoding versus a cipher
An encoding changes representation so data can be stored or transported. Base64 is an example; it does not require a secret. A cipher transforms a message using a rule or key. In a Caesar cipher, shift every letter the same number of places: A → D for shift 3. To undo it, shift backward. ROT13 is Caesar with shift 13. Wrap around at Z.

### Lesson 3 — Recognize, test, verify
Look at the alphabet and grouping. Eight-bit groups suggest bytes; pairs of hex digits suggest hex; dots and dashes suggest Morse. These are clues, not guarantees. Try one transformation, inspect the result, and preserve the intermediate text. Multiple layers are possible: Base64 → Caesar → a room instruction.

## Worked example: Caesar shift
`WDEOH` was encrypted by shifting letters forward three places. Undo the shift to recover a location.

```python
def caesar(text, shift):
    result = []
    for char in text:
        if "A" <= char <= "Z":
            result.append(chr((ord(char)-65+shift)%26+65))
        elif "a" <= char <= "z":
            result.append(chr((ord(char)-97+shift)%26+97))
        else:
            result.append(char)
    return "".join(result)

print(caesar("WDEOH", -3))
```
